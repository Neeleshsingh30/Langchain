```python
import sys
!{sys.executable} -m pip install -U langchain-text-splitters
from langchain_text_splitters import CharacterTextSplitter
```

```python
structured_data = """ID:101 | Name: Rahul | Role: Developer\n\nID:102 | Name: Neha | Role: Tester\n\nID:103 | Name: Aman | Role: Manager"""
```

```python
unstructured_data = "Generative AI is a subset of deep learning. It uses transformer models to create text. Students love learning LangChain because it makes things easy."
```

```python
# Splitter Configuration
char_splitter = CharacterTextSplitter(
    separator="\n\n",
    chunk_size=40,   
    chunk_overlap=0
)
```

```python
print("=== CASE 1: Structured Data (it's work) ===")
chunks_success = char_splitter.split_text(structured_data)
for i, chunk in enumerate(chunks_success):
    print(f"Chunk {i+1} (Size {len(chunk)}): {chunk}")
```

```python
chunks_success
```

```python
# Output: The data will be cleanly split into 3 chunks, all under 40 characters.
```

```python
print("\n=== CASE 2: Unstructured Data (Fail ) ===")
chunks_fail = char_splitter.split_text(unstructured_data)
for i, chunk in enumerate(chunks_fail):
    print(f"Chunk {i+1} (Size {len(chunk)}): {chunk}")
```

```python
# Output: It generates only 1 long chunk of size 147. The chunk limit is broken because it couldn't find the '\n\n' separator!
```
